Read in the event permit csvs and concatenate to 1 df

```python
# Read in the event permit csvs and concatenate to 1 df
event_permit_2014_2018 = pd.read_csv('NYC_Event_Permits_2014_2018.csv')
event_permit_2019_2020 = pd.read_csv('NYC_Event_Permits_2019_2020.csv')
event_permit_2021_2021 = pd.read_csv('NYC_Event_Permits_2021_2021.csv')
event_permit_2022_2022= pd.read_csv('NYC_Event_Permits_2022_2022.csv')
event_permit_2023_2023 = pd.read_csv('NYC_Event_Permits_2023_2023.csv')
event_permit_2024_2024 = pd.read_csv('NYC_Event_Permits_2024_2024.csv')
```

```python
event_permit_df = pd.concat([event_permit_2014_2018, event_permit_2019_2020, event_permit_2021_2021,
                             event_permit_2022_2022, event_permit_2023_2023, event_permit_2024_2024], ignore_index=True)
```

```python
event_permit_df.to_csv('event_permit_full.csv', index=False)
```
